```python
df['Order Date']=pd.to_datetime(df['Order Date'])
```

```python
df['year']=df['Order Date'].dt.year
```

```python
Total_Revenue=df['Total Sales'].sum()
print("Total Revenue:",Total_Revenue)
```

```python
Monthly_sales =df.groupby('month')['Total Sales'].sum()
print(Monthly_sales)

```

```python
monthly_sales = df.groupby('month')['Total Sales'].sum()

highest_month = monthly_sales.idxmax()
highest_revenue = monthly_sales.max()

print("Highest Revenue Month:", highest_month)
print("Highest Revenue:", highest_revenue)
```

```python
Category_sales=df.groupby('Category')['Total Sales'].sum()
print(Category_sales)
```

```python
Highest_category=Category_sales.idxmax()
Highest_sales=Category_sales.max()
print("Top category:",Highest_category)
print("Total Revenue:",Highest_sales)
```

```python
Region_sales=df.groupby('Region')['Total Sales'].sum()
print(Region_sales)
```

```python
Top_Region=Region_sales.idxmax()
Top_Revenue=Region_sales.max()
print("Top Region:",Top_Region)
print("Top Revenue:",Top_Revenue)
```

```python
customer_sales = df.groupby('Customer ID')['Total Sales'].sum()

print(customer_sales.head(10))
```

```python
top_10_customers = customer_sales.sort_values(ascending=False).head(10)

print(top_10_customers)
```

```python
payment_counts = df['Payment Method'].value_counts()

print(payment_counts)
```

```python
most_popular_payment = payment_counts.idxmax()
highest_count = payment_counts.max()

print("Most Popular Payment Method:", most_popular_payment)
print("Number of Transactions:", highest_count)
```

```python
average_order_value = df['Total Sales'].mean()

print("Average Order Value:", average_order_value)
```

```python
product_sales = df.groupby('Product')['Total Sales'].sum()

print(product_sales)
```

```python
top_product = product_sales.idxmax()
top_product_revenue = product_sales.max()

print("Top Product:", top_product)
print("Total Revenue:", top_product_revenue)
```

```python
gender_sales = df.groupby('Gender')['Total Sales'].sum()

print(gender_sales)
```

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0, 20, 30, 40, 50, 100],
    labels=['Under 20', '20-30', '31-40', '41-50', 'Above 50']
)

age_sales = df.groupby('Age Group', observed=False)['Total Sales'].sum()

print(age_sales)
```

```python
import matplotlib.pyplot as plt

age_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Revenue by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
Category_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='Category',
    y='Total Sales',
    estimator=sum,
    errorbar=None
)

plt.title('Total Revenue by Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Total Sales', bins=20, kde=True)

plt.title('Sales Distribution')
plt.xlabel('Total Sales')
plt.ylabel('Number of Orders')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x='Gender', y='Total Sales')

plt.title('Sales Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.show()
```

```python
correlation = df[['Age', 'Quantity', 'Unit Price', 'Total Sales']].corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Retail Sales Correlation Heatmap')
plt.show()
```

```python
daily_sales = df.groupby('Order Date')['Total Sales'].sum()

print(daily_sales.head(10))
```

```python
plt.figure(figsize=(12, 5))

daily_sales.plot(kind='line')

plt.title('Daily Sales Trend')
plt.xlabel('Order Date')
plt.ylabel('Total Revenue')
plt.grid(True)
plt.show()
```
